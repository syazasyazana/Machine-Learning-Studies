```python
from scipy import stats

# Accuracies of each machine learning model
accuracies = [0.69, 0.75, 0.70, 0.72, 0.91]
model_names = ['L1 Regularization', 'XGBoost', 'CatBoost', 'LightGBM', 'FastTree']

# Perform t-tests for all pairs of models
for i in range(len(accuracies)):
    for j in range(i+1, len(accuracies)):
        # Create lists of repeated accuracies with the same length
        model1_accuracies = [accuracies[i]] * 10000
        model2_accuracies = [accuracies[j]] * 10000
        
        # Perform t-test
        t_statistic, p_value = stats.ttest_ind(model1_accuracies, model2_accuracies, nan_policy='omit')
        
        # Print results
        print(f"T-test between {model_names[i]} and {model_names[j]}:")
        print(f"T-statistic: {round(t_statistic, 4)}")  
        print(f"P-value: {p_value:.4f}")
        
        # Determine conclusion
        if p_value < 0.05:
            print("Conclusion: Reject null hypothesis (significant difference)")
        else:
            print("Conclusion: Fail to reject null hypothesis (no significant difference)")
        print()
```
